# data visualization 

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x='diagnosis', data=df)
plt.title('Benign vs Malignant')
plt.show()
```

```python
sns.histplot(df['radius_mean'], kde=True)
plt.title('Radius Mean Distribution')
plt.show()
```

```python
sns.boxplot(x='diagnosis', y='radius_mean', data=df)
plt.title('Radius Mean by Diagnosis')
plt.show()
```

```python
sns.boxplot(x='diagnosis', y='area_mean', data=df)
plt.show()
```

```python
sns.boxplot(x='diagnosis', y='perimeter_mean', data=df)
plt.show()
```

# correlation heatmap

```python
plt.figure(figsize=(15,10))
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.scatterplot(data=df, x='radius_mean', y='area_mean', hue='diagnosis')
plt.title('Radius Mean vs Area Mean')
plt.show()
```
